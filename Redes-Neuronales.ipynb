```python
# Tarea 7
# Autor: Carlos Rios

# Creando en entorno virtual
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

df = pd.read_excel(r"C:/base-de-datos-violencia-intrafamiliar-ano-2024_v3.xlsx")

# Convertir VIC_SEXO a 0/1
y = df["VIC_SEXO"].replace({1:0, 2:1}).astype(int)
print("Valores únicos de Y:", y.unique())
```

```python
# PASO 2 — Definir las variables finales

vars_finales = [
 'AGR_SEXO', 'VIC_TRABAJA', 'VIC_EDAD', 'VIC_REL_AGR', 'HEC_TIPAGRE',
 'AGR_TRABAJA', 'NUM_HIJ_HOM', 'NUM_HIJ_MUJ', 'TOTAL_HIJOS',
 'VIC_EST_CIV', 'VIC_ESCOLARIDAD', 'INST_DENUN_HECHO', 'VIC_DISC',
 'AGR_EST_CIV', 'AGR_OTROS_N_OS', 'AGR_OTRAS_N_AS', 'AGR_OTRAS_MUJ',
 'AGR_OTROS_HOM', 'AGRESORES_OTROS_TOTAL'
]

```

```python
# PASO 3 — Construir X

X = df[vars_finales].copy()
X = X.fillna(X.mean())

print("Columnas reales usadas por X:", list(X.columns))
print("Total columnas X =", len(X.columns))
```

```python
# PASO 4 — Normalizar

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# PASO 5 — Train/Test split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
# PASO 6 — Construir el modelo

modelo = models.Sequential([
    tf.keras.Input(shape=(X_train.shape[1],)),
    layers.Dense(32, activation="relu"),
    layers.Dense(16, activation="relu"),
    layers.Dropout(0.2),
    layers.Dense(1, activation="sigmoid")
])

modelo.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
# PASO 7 - Entrenar el modelo

print("Entrenando modelo...")

hist = modelo.fit(
    X_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
# PASO 8 - Evaluar el modelo

loss, acc = modelo.evaluate(X_test, y_test, verbose=0)
print(f"\nExactitud en TEST: {acc:.4f}")
```

```python
# Prueba de predicción

# Escenario aleatorio
escenario = np.array([
    1, 0, 34, 1, 1122, 1, 1, 0, 1,
    3, 2, 1, 0, 2, 0, 0, 0, 0, 0
]).reshape(1, -1)

# Escalar usando el scaler ya entrenado
escenario_scaled = scaler.transform(escenario)

# Realizar predicción
pred = modelo.predict(escenario_scaled)[0][0]

print("Probabilidad predicha (sigmoid):", pred)

if pred >= 0.5:
    print("🔷 Predicción: Víctima hombre (1)")
else:
    print("🔶 Predicción: Víctima mujer (0)")
```
